```
#Write a program to read the radius of a circle and print its area and circumference.
import math
radius = 10

area_of_circle = math.pi * radius * pow(radius, 2)
print(f"Area of circle is: {round(area_of_circle, 2)}")

cirumf = 2* math.pi * radius
print(f"Circumference of circle is: {round(cirumf, 2)}")
```

```
#Write a program to read the length and breadth of a rectangle and print its area and perimeter.
length_rectangle = 10
width_rectangle = 20

area_of_rectangle = length_rectangle * width_rectangle
print(f"Area of rectangle is: {area_of_rectangle}")

peri = 2*(length_rectangle + width_rectangle)
print(f"Perimeter of rectangle is: {peri}")
```

```
#Write a program to swap two numbers using a third variable.
x = 11
y = 20
z = 33

z = x
x = y
y = z

print(x, y)
```

```
#Write a program to swap two numbers without using a third variable.
n = 22
m = 88

n = n + m
m = n - m
n = n - m

# or x, y = y,x - gets it swapped as well

print(n, m)
```

```
#Write a program to read a temperature in Celsius and convert it to Fahrenheit.

temp_in_celsius = 50
temp_in_f = (temp_in_celsius * 1.8) + 32

print(temp_in_f)
```

```
#Write a program to read the marks of 5 subjects and print the total and average.

maths = 11
english = 30
science = 20
french = 10
computers = 28

total = maths + english + science + french + computers
average = (maths + english + science + french + computers)/4

print(f"The total marks are: {total}, and the average marks are: {average}")
```

```
#Write a program to read seconds and convert them into hours, minutes and seconds.

the_second = 12020

hours = the_second // 3600
leftover_seconds = the_second % 3600
mins = leftover_seconds // 60
seconds = leftover_seconds % 60

print(f"{the_second} is {hours} hours, {mins} minutes, {seconds} seconds")
```
