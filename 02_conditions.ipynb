```python
#Write a program to read a number and check whether it is even  or odd.
num = 4

if num % 2 == 0:
    print("even number")
    print(f"{num} is even")
elif num % 2 != 0:
    print("odd number")
    print(f"{num} is odd")
else:
    print("Not a number")
```

```python
#Write a program to read a number and check whether it is positive, negative or zero.

num = 0

if num < 0:
    print(f"{num} is negative")
elif num > 0:
    print(f"{num} is positive")
elif num == 0:
    print(f"{num} is zero")
else:
    print(f"{num} is undefined")
```

```python
#Write a program to read a character and check whether it is a vowel or a consonant.
char = "y"

if char == "a" or char == "e" or char == "i" or char == "o" or char == "u" :
    print(f"{char} is a vowel")
elif char == "A" or char == "E" or char == "I" or char == "O" or char == "U":
    print(f"{char} is a vowel")
else:
    print(f"{char} is not a consonant")
```

```python
#Write a program to read a character and check whether it is an alphabet, digit or special symbol.
character = "navn33"

if character.isalpha():
    print(F"{character} is an alphabetical character")
elif character.isdigit():
    print(F"{character} is a digit")
elif character.isalnum():
    print("It contains both alphabets and digits.")
else:
    print(F"{character} is a special symbol")
```

```python
#Write a program to read the marks of a student and print the grade (A/B/C/D/Fail).
marks = 77

if marks == 100 or marks >= 90  :
    print(f"Your marks is {marks}. You got an A grade")
elif marks < 90 or marks >= 70:
    print(f"Your marks is {marks}. You got an B grade")
elif marks < 70 or marks >= 50:
    print(f"Your marks is {marks}. You got an C grade")
elif marks < 50 or marks >= 20:
    print(f"Your marks is {marks}. You got an D grade")
elif marks > 100:
    print("Thats not correct")
else:
    print(f"Your marks is {marks}. You are Fail")
```
